Given two lists, find the missing element from the second list, ignoring multiple missing elements.

```java
public class Solution1 {
    public static int findMissingElement(int[] arr1, int[] arr2) {
        int sum1 = 0, sum2 = 0;
        for (int num : arr1) sum1 += num;
        for (int num : arr2) sum2 += num;
        return sum1 - sum2;
    }

    public static void main(String args[]){
        int arr1[] = {1, 2, 3, 4, 5};
        int arr2[] = {1, 2, 4, 5};
        int missing = findMissingElement(arr1, arr2);
        System.out.println(missing);
    }
}
```

```java
Solution1.main(null);
```

Q. Given a string that may contain numeric and special characters, check if it is a palindrome, ignoring those characters.

```java
public class Solution2 {
    public boolean isPalindrome(String s) {
        StringBuilder filtered = new StringBuilder();
        for (char c : s.toCharArray()) {
            if (Character.isLetterOrDigit(c)) {
                filtered.append(Character.toLowerCase(c));
            }
        }
        int left = 0, right = filtered.length() - 1;
        while (left < right) {
            if (filtered.charAt(left) != filtered.charAt(right)) return false;
            left++;
            right--;
        }
        return true;
    }

    public static void main(String[] args) {
        Solution2 sol = new Solution2();

        String s = "A man, a plan, a canal: Panama";

        System.out.println(sol.isPalindrome(s));
    }
    
}
```
