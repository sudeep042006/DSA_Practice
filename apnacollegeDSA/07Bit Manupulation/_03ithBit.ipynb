```java
//clear ith Bit
Scanner sc = new Scanner(System.in);
int m = sc.nextInt();
int n = sc.nextInt();
int bitmask = ~(1<<n);
System.out.println((m & bitmask));
String binary = Integer.toBinaryString(m & bitmask);
System.out.println(binary);
```

```java
//update ith bit
Scanner sc = new Scanner(System.in);
int n = sc.nextInt();
int i = sc.nextInt();
int value = sc.nextInt(); // 0 or 1

// Clear the bit first
n = n & ~(1 << i);

// Set it if value is 1
n = n | (value << i);

System.out.println(n);
```
